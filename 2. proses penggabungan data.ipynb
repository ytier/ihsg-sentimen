```python
import pandas as pd
from pathlib import Path


def bersihkan_dan_gabungkan_data(file1, file2, file_output):
    print("Memulai proses pembacaan dan pembersihan data...")

    # 1. Membaca kedua file
    df1 = pd.read_csv(file1)
    df2 = pd.read_csv(file2)

    # 2. Menggabungkan data
    df_combined = pd.concat([df1, df2], ignore_index=True)

    # 3. Pembersihan multi-layer
    # Layer A: Hapus baris yang kolom teksnya kosong
    df_cleaned = df_combined.dropna(subset=["text"])
    # Layer B: Hapus duplikat ID
    df_cleaned = df_cleaned.drop_duplicates(subset=["id"], keep="first")
    # Layer C: Hapus duplikat teks eksak (indikasi bot/spam)
    df_cleaned = df_cleaned.drop_duplicates(subset=["text"], keep="first")

    # 4. Manajemen direktori output
    output_path = Path(file_output)
    if output_path.parent:
        output_path.parent.mkdir(parents=True, exist_ok=True)

    # 5. Simpan hasil
    df_cleaned.to_csv(output_path, index=False)

    # Laporan
    print("\n--- LAPORAN HASIL PEMBERSIHAN ---")
    print(f"Total baris mentah gabungan : {len(df_combined)}")
    print(f"Total baris valid akhir     : {len(df_cleaned)}")
    print(f"Total data kotor dieleminasi: {len(df_combined) - len(df_cleaned)}")
    print(f"Proses selesai. File tersimpan di: {output_path.resolve()}")


if __name__ == "__main__":
    # File sumber (G:/My Drive/...)
    file_pertama = r"G:\My Drive\skripsi\proje\proje\data\raw\dataset_ihsg_murni.csv"
    file_kedua = r"G:\My Drive\skripsi\proje\proje\data\raw\dataset_ihsg_murni 2.csv"

    # Hasil gabungan disimpan di folder yang sama (raw)
    file_hasil = (
        r"G:\My Drive\skripsi\proje\proje\data\raw\dataset_ihsg_final_bersih.csv"
    )

    bersihkan_dan_gabungkan_data(file_pertama, file_kedua, file_hasil)
```
